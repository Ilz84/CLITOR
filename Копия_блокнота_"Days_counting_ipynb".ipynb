```python
from datetime import datetime

def get_days_diff():
    while True:
        user_input = input("\nВведите дату в формате ГГГГ-ММ-ДД (или 'q' для выхода): ").strip()

        # Выход из программы
        if user_input.lower() == 'q':
            print("Программа завершена.")
            break

        try:
            # Парсим дату от пользователя
            input_date = datetime.strptime(user_input, "%Y-%m-%d").date()
            today = datetime.today().date()

            # Вычисляем разницу
            delta = today - input_date
            days_passed = delta.days

            if days_passed < 0:
                print(f"Эта дата ещё не наступила. До неё осталось {abs(days_passed)} дней.")
            elif days_passed == 0:
                print("Это сегодня!")
            else:
                print(f"С указанной даты прошло {days_passed} дней.")

        except ValueError:
            print("Ошибка: неверный формат даты или несуществующая дата. Пожалуйста, используйте формат ГГГГ-ММ-ДД (например, 2024-05-20).")
        except Exception as e:
            # Перехватываем любые другие непредвиденные ошибки
            print(f"Произошла непредвиденная ошибка: {e}")

        # Спрашиваем, хочет ли пользователь сделать новый расчёт
        again = input("Хотите сделать новый расчёт? (да/нет): ").strip().lower()
        if again not in ('да', 'д', 'yes', 'y'):
            print("Программа завершена.")
            break

if __name__ == "__main__":
    get_days_diff()
```
